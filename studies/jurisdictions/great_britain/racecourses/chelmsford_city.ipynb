## Racecourse identity

**Governed racecourse identity:** `Chelmsford City`

Chelmsford City contains its established Polytrack all-weather course and a separately constructed turf track inside it. The BHA approved the turf course for Flat racing in 2018, and official racecourse material later described the turf track as fully railed and the site as home to both all-weather and turf tracks. The exact first competitive turf-race activation date remains unresolved.

Evidence, Inside Rails derivation, candidates and unresolved questions are kept separate.
