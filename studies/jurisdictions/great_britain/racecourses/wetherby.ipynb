## Racecourse identity

**Governed racecourse identity:** `Wetherby`

Wetherby is governed as one racecourse with separately recognised Hurdles' and Chase courses. BHA evidence remeasures and describes the Hurdle and Chase courses separately, and current Wetherby material states that Flat races are run on the Hurdles' course. Flat racing began in 2015 and does not create a third permanent course.

Evidence, Inside Rails derivation, candidates and unresolved questions are kept separate.
