## Revised course/track model

BHA evidence distinguishes the post-redevelopment **Straight Course** and **Round Course**.

The old Jump model required correction. A 1999 BHB release genuinely describes the Hurdle Course as one of two National Hunt tracks, but that predates the 2004–06 redevelopment. During redevelopment in 2005, Ascot Course Manager David Warner described **two tracks overall**: the Flat track and a combined **steeplechase and hurdle track**, in a section headed **National Hunt Course**. Technical reporting in 2011 again uses singular National Hunt course/track language, and in 2020 Ascot's Clerk of the Course explained that chase and hurdle races run inside the round Flat course on the National Hunt (jumps) track.

For 2015–2026, the governed top-level inventory is therefore **Straight Course, Round Course, National Hunt Course**. Hurdle and chase/steeplechase are uses/alignments on the National Hunt track, not separate top-level entities.

## Course layout, Old Mile and historical resolution

Post-redevelopment evidence places the National Hunt/jumps track **inside the round Flat course** and documents seasonal rail/alignment changes.

At The Races names an **Old Mile Course** which joins the Round Course in Swinley Bottom. This establishes a named Flat route and its relationship to the Round Course, but no BHA/Ascot source found in this pass settles whether it should be governed as a separate constituent course or as a branch/start configuration.

Historical anchors:
- **1999:** two National Hunt tracks in the pre-redevelopment layout.
- **2004–06:** Straight Mile rebuilt/moved 42m north; substantial National Hunt/chase reconstruction.
- **2005:** Ascot Course Manager describes Flat track + combined steeplechase/hurdle track.
- **2011:** singular National Hunt track/course in technical grounds reporting.
- **2015:** BHA measured `STEEPLECHASE`, `HURDLE`, `NHFR` distance sets.
- **2020:** Ascot Clerk of the Course describes chase/hurdle racing on the singular National Hunt track inside the round Flat course.

Thus the 1999 two-NH-track model is retained historically, not carried into 2015–2026.
