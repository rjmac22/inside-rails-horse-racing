## Recognised courses and tracks

Primary BHA evidence explicitly distinguishes Ascot's **straight course** and **round course** in Flat racing.

A separate BHA professional remeasurement of British Jump racecourses, implemented from 1 June 2015, lists distinct Ascot distance sets under **STEEPLECHASE** and **HURDLE**. That is sufficient to treat the steeplechase and hurdle routes as separately recognised racing configurations for this first-pass inventory.

The BHA source does **not** establish exact permanent formal names equivalent to `Steeplechase Course` and `Hurdle Course` for Ascot. The notebook therefore uses descriptive working labels for those two configurations rather than inventing formal names.

## Surface

For the two Flat configurations, current first-pass classification is **Turf**. Ascot's own current raceday material depicts/describes racing on a green grass racecourse, which provides direct venue evidence rather than inferring the surface merely from the racing code.

For the separately identified Jump configurations, surface remains **unresolved in this notebook** pending a source that explicitly establishes the surface at the required configuration level. It is not filled simply because Ascot is known generally as a turf venue.

## Handedness

- **Straight Course** — `Not applicable`. A wholly straight route has no meaningful left/right handedness.
- **Round Course** — `Right-handed`. The BHA explicitly classified Ascot as right-handed because the major bend is right-handed when races are run around a turn.
- **Steeplechase configuration** — unresolved at configuration level in this first pass.
- **Hurdle configuration** — unresolved at configuration level in this first pass.

The BHA's racecourse-level/Flat handedness classification is not automatically copied onto separately evidenced Jump configurations.

## Racing use

- **Straight Course** — Flat.
- **Round Course** — Flat.
- **Steeplechase configuration** — Steeplechase / chase.
- **Hurdle configuration** — Hurdle.

Ascot's own current programme confirms that the venue stages both Flat and Jumps racing.


## Course layout and physical characteristics

The current first pass establishes only the distinctions required to identify the four supported racing configurations above.

No further permanent sub-courses, dimensions, gradients, exact start geometry or rail-position variants are promoted here without separate evidence.

The Flat distinction is particularly important: the BHA has previously been able to move Ascot races between the round and straight course in response to ground conditions, demonstrating that these are operationally distinct racing configurations rather than alternative names for the same route.


## Historical changes

The BHA Jump remeasurement evidence became operational on **1 June 2015**, at the beginning of the study period, and separately records Ascot steeplechase and hurdle distance sets.

This notebook does not yet establish when the current Flat or Jump configurations first came into existence, nor whether their physical geometry changed materially within the 2015–2026 study period. Exact temporal boundaries remain unresolved unless explicitly evidenced.
