## Recognised courses and tracks

### Flat

BHA evidence explicitly distinguishes Ascot's **Straight Course** and **Round Course**. A 2012 contingency plan treated the two as operationally distinct: round-course races were at risk while straight-course racing could continue.

### Jump

A 1999 BHB release called the hurdle course the wetter of **two National Hunt tracks at Ascot** and described an all-steeplechase contingency if the hurdle course was unusable. The 2015 BHA professional remeasurement then lists separate `STEEPLECHASE` and `HURDLE` distance schedules.

This supports two National Hunt track entities. The exact permanent name of the non-hurdle track is not established by the primary evidence reviewed, so **Steeplechase Course (working label)** is used provisionally.

The same 2015 schedule also contains `NHFR`. That proves **NH Flat racing use**, not a fifth physical course.

## Evidence and provenance

Each assertion is recorded separately. Course identity does not automatically prove surface, handedness or use; distance headings do not automatically prove course identity; secondary names are not silently promoted.

Where Inside Rails makes a derivation, the **external publisher remains the source authority** and the derivation is identified in `verification_status` / `evidence_note`.


```python
BHA = "British Horseracing Authority"
BHB = "BHB / BHA archive"
ASCOT = "Ascot Racecourse"
U_FLAT = "https://www.britishhorseracing.com/press_releases/ascot-may-11th-and-12th-extra-race-and-divisions-to-safeguard-racing/"
U_BHB = "https://www.britishhorseracing.com/press_releases/bhb-announces-additional-contingency-arrangements-to-maximise-racing-prospects-for-ascot-this-friday/"
U_JUMP = "https://www.britishhorseracing.com/wp-content/uploads/2015/05/Jump-distance-changes-1-June-2015.pdf"
U_SURFACE = "https://www.ascot.com/racedays/summer-mile-family-raceday"
U_HAND = "https://www.britishhorseracing.com/press_releases/new-range-of-flat-race-starting-procedures-to-be-trialled/"
T_FLAT = "ASCOT, MAY 11TH AND 12TH: EXTRA RACE AND DIVISIONS TO SAFEGUARD RACING"
T_BHB = "BHB ANNOUNCES ADDITIONAL CONTINGENCY ARRANGEMENTS TO MAXIMISE RACING PROSPECTS FOR ASCOT THIS FRIDAY"
T_JUMP = "Changes to Jump race distances from 1 June 2015"

def prov(course, characteristic, value, authority, title, url, note, status="verified", valid_from=null):
    return ["Great Britain","Ascot",course,characteristic,value,valid_from,null,authority,title,url,"2026-08-10",note,status]

course_characteristic_provenance = pd.DataFrame([
    prov("Straight Course","course_identity","Straight Course",BHA,T_FLAT,U_FLAT,"BHA distinguishes straight and round courses operationally."),
    prov("Round Course","course_identity","Round Course",BHA,T_FLAT,U_FLAT,"BHA distinguishes round and straight courses operationally."),
    prov("Hurdle Course","course_identity","Hurdle Course",BHB,T_BHB,U_BHB,"Official release names hurdle course and says it is one of two National Hunt tracks."),
    prov("Steeplechase Course (working label)","course_identity","separate steeplechase/chase track; canonical name unresolved",BHB,T_BHB,U_BHB,"Two National Hunt tracks plus all-steeplechase contingency support a separate non-hurdle track.","verified_entity_name_unresolved"),
    prov("Hurdle Course","study_period_presence","HURDLE distance set",BHA,T_JUMP,U_JUMP,"BHA 2015 survey lists HURDLE distances for Ascot.","verified","2015-06-01"),
    prov("Steeplechase Course (working label)","study_period_presence","STEEPLECHASE distance set",BHA,T_JUMP,U_JUMP,"BHA 2015 survey lists STEEPLECHASE distances for Ascot.","verified_entity_name_unresolved","2015-06-01"),
    prov("Straight Course","surface","Turf",ASCOT,"Racing To Zero Summer Mile Family Raceday",U_SURFACE,"Official Ascot material supplies first-pass grass/turf evidence.","verified_first_pass"),
    prov("Round Course","surface","Turf",ASCOT,"Racing To Zero Summer Mile Family Raceday",U_SURFACE,"Official Ascot material supplies first-pass grass/turf evidence.","verified_first_pass"),
    prov("Straight Course","handedness","Not applicable",BHA,T_FLAT,U_FLAT,"BHA establishes a Straight Course. Inside Rails derives handedness = Not applicable because a wholly straight route has no left/right turn.","derived_from_verified_structure"),
    prov("Round Course","handedness","Right-handed",BHA,"NEW RANGE OF FLAT RACE STARTING PROCEDURES TO BE TRIALLED",U_HAND,"BHA says Ascot is right-handed due to the major bend when racing around a turn."),
    prov("Straight Course","primary_use","Flat",BHA,T_FLAT,U_FLAT,"BHA schedules Flat races on straight course."),
    prov("Round Course","primary_use","Flat",BHA,T_FLAT,U_FLAT,"BHA schedules Flat races on round course."),
    prov("Hurdle Course","primary_use","Hurdle",BHB,T_BHB,U_BHB,"Official release distinguishes the hurdle course."),
    prov("Steeplechase Course (working label)","primary_use","Steeplechase / chase",BHB,T_BHB,U_BHB,"All-steeplechase contingency supports steeplechase use on the other NH track.","verified_entity_name_unresolved"),
    prov(null,"racing_use","NH Flat",BHA,T_JUMP,U_JUMP,"BHA schedule contains NHFR row; physical-course assignment unresolved.","verified_use_course_unresolved","2015-06-01"),
], columns=["jurisdiction","racecourse_identity","course_or_track_name","characteristic","value","valid_from","valid_to","source_authority","source_title","source_url","accessed_date","evidence_note","verification_status"])
course_characteristic_provenance
```
