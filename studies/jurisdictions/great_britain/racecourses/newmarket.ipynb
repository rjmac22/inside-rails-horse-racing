## Scope

- Jurisdiction: **Great Britain**
- Study period: **2015-01-01 to 2026-05-27**
- Source labels: `Newmarket`, `Newmarket (July)`
- Official racecourses: **Newmarket — Rowley Mile** and **Newmarket — July Course**
- Key conclusion: The Jockey Club explicitly states that Newmarket has two racecourses — the Rowley Mile and the July Course. The source's July-specific label is therefore retained as the July Course identity; plain `Newmarket` is resolved to the Rowley Mile by the observed Source Version 1 naming convention.

## Racecourse identity

The earlier analytical parent `Newmarket` is not retained as a racecourse identity. Official Jockey Club material calls the Rowley Mile and July Course Newmarket's **two racecourses**, and gives them separate visitor/racecourse locations.

For Source Version 1 the governed mapping is:

- `Newmarket` → **Newmarket — Rowley Mile** (`source_label_convention`)
- `Newmarket (July)` → **Newmarket — July Course** (`explicit_source_label`)

This convention is bounded to Source Version 1 and must be revalidated for future source versions rather than assumed forever.

### Source-label resolution evidence
