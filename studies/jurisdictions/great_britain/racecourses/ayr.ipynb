## Racecourse identity

**Governed racecourse identity:** `Ayr`

Ayr is governed as one racecourse containing two peer physical racing-course identities: **Flat Course** and **Jumps Course**.

BHA evidence distinguishes the Flat track from the Hurdles/Jumps course by transferring Flat fixtures to the Hurdles Course when the Flat track was unavailable. Ayr Racecourse's own history also states that the jumps course was established in 1950.

Ayr Racecourse separately refers to the six-furlong **straight course**. For Study 03 it is retained as a named physical component/route of the Flat Course rather than promoted to a third peer course identity, because the available official material does not place it at the same hierarchy level as the Flat and Jumps courses.
