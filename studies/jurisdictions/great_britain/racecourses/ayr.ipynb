## Racecourse identity

**Governed racecourse identity:** `Ayr`

Ayr is governed as one racecourse containing separate Flat and Jumps racing courses. BHA study-period evidence distinguishes Ayr's Flat track from its Hurdles/Jumps course by transferring Flat fixtures onto the Hurdles Course when the Flat track was unavailable. The wide straight sprint section is retained as a lower-level component/candidate rather than promoted to a third top-level course.

Evidence, Inside Rails derivation, candidates and unresolved questions are kept separate.
