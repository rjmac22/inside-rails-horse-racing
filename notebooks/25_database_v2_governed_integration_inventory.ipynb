```python
from pathlib import Path
import sqlite3

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

V1_DB = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "database"
    / "releases"
    / "inside_rails_v1.sqlite3"
)

# Fail immediately if the accepted immutable Database v1 release is not where
# the governed repository structure says it should be. We do not guess or fall
# back to another database because this stage is specifically reconciling v2
# design against the accepted v1 physical schema.
assert V1_DB.exists(), f"Accepted Database v1 not found: {V1_DB}"

# Open Database v1 strictly read-only. This cell is inspection only and must
# not create tables, indexes, views or any other state inside an accepted release.
connection = sqlite3.connect(f"file:{V1_DB}?mode=ro", uri=True)

try:
    # Inventory the existing user-defined tables and views first. The v2 table
    # design should extend the accepted architecture rather than inventing
    # parallel names or accidentally duplicating an existing structure.
    objects = connection.execute(
        """
        SELECT type, name
        FROM sqlite_master
        WHERE type IN ('table', 'view')
          AND name NOT LIKE 'sqlite_%'
        ORDER BY type, name
        """
    ).fetchall()

    print("Accepted Database v1 objects:")
    for object_type, name in objects:
        print(f"  {object_type:5}  {name}")

    # Inspect the declared columns of every physical table so the next design
    # stage can identify which v1 grains and keys are carried forward unchanged
    # and which Notebook 04–22 outputs genuinely require new structures.
    print("\nTable grains / columns:")
    for object_type, name in objects:
        if object_type != "table":
            continue

        columns = connection.execute(
            f'PRAGMA table_info("{name}")'
        ).fetchall()

        print(f"\n{name}")
        for column in columns:
            (
                _,
                column_name,
                declared_type,
                not_null,
                default_value,
                primary_key,
            ) = column

            # Make key and nullability constraints visible in the notebook
            # output because they are material to deciding whether a v2
            # extension belongs in an existing table or a new attached table.
            markers = []
            if primary_key:
                markers.append("PK")
            if not_null:
                markers.append("NOT NULL")

            marker_text = f" [{' | '.join(markers)}]" if markers else ""
            print(f"  - {column_name}: {declared_type}{marker_text}")
finally:
    # Close the read-only connection explicitly so this inspection stage leaves
    # no lingering handle on the immutable accepted release.
    connection.close()
```

## Notebook 19 horse and pedigree identity — proposed exact field design

Notebook 19 requires three Database v2 structures because it established three distinct relational grains:

1. a governed provisional horse occurrence;
2. the assignment of applicable source-backed runner rows to those occurrences;
3. the governed pedigree/identity decisions that determine whether adjacent histories remain one horse or split into different horses.

The three physical tables are:

- `identity_horse_occurrence`;
- `identity_runner_horse_occurrence`;
- `identity_horse_pedigree_decision`.

The governing principle remains:

> raw `horse`, `sire`, `dam` and `damsire` are immutable source assertions; provisional identity and governed pedigree decisions are additive analytical structures.

A provisional horse occurrence is source-internal only.

It is **not**:

- an official registration number;
- a life number;
- a provider-independent horse identity;
- evidence that the same label is unique outside Source Version 1.

# 19. Governance-release lineage

Database v2 must preserve the distinction between structural Database v1 governance and new Database v2 governance.

The carried-forward structural rows remain governed by the release that established their structural meaning.

Therefore, existing:

- `core_source_race_occurrence`;
- `core_runner_participation`;

do not need their historical `governance_release_id` rewritten merely because they appear inside the Database v2 file.

New Database v2 structures point to the new Database v2 governance release.

The Database v2 import manifest then records the complete Database v2 build and its relationship to the prior accepted Database v1 release.

This preserves:

> v1 structural lineage + v2 additive semantic governance

rather than pretending every carried-forward structural fact was first created in v2.

# 20. Foreign-key deletion policy

Accepted database releases are immutable.

Database v2 therefore has no requirement for cascading destructive updates.

New foreign keys should use ordinary fail-closed referential behaviour.

Do not use:

- `ON DELETE CASCADE`;
- `ON UPDATE CASCADE`;

to make governance evidence, source lineage or identities disappear automatically.

A broken reference should fail construction or validation.

It should not trigger automatic deletion of dependent evidence.

# 21. Correction provenance rule

A governed analytical value may differ from the raw source only where an accepted rule authorises it.

The physical relationship should use the narrowest applicable provenance structure.

Examples:

### Notebook 17 sex correction

Runner analytical value may differ from raw `sex`.

The runner extension retains:

- corrected analytical value;
- correction status;
- exact verification reference.

Raw `sex` remains available through the immutable source record.

### Notebook 20 connection supplementation

Runner governed jockey/trainer/owner value may differ from raw blank.

The runner extension links to the applicable:

`governance_connection_value_decision`.

The connection decision then links to:

`governance_manual_verification`.

### Notebook 19 pedigree correction

The identity layer links a transition to the applicable:

`governance_horse_pedigree_specialist_decision`.

Raw pedigree assertions remain source-backed.

### Unresolved evidence

No replacement value is created.

The evidence remains stored while the analytical value remains null or raw-only as governed.

# 22. Correction candidates remain candidates

A verification with:

`database_action = 'source_correction_candidate'`

does not automatically become an accepted Database v2 corrected value.

This applies to currently retained candidate states such as bounded Notebook 14, 15 and 16 contradictions unless a specific durable implementation promotes the correction.

The database may preserve:

- the raw value;
- the verification;
- the candidate action;
- the contradiction state.

It must not silently promote the candidate into a canonical value merely because external evidence suggests the source is wrong.

Accepted corrections, such as the exact governed Notebook 17 sex corrections and current Notebook 19 pedigree corrections, remain usable because their specific integration contracts authorise that analytical treatment.
